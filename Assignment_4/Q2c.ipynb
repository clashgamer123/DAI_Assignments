Load all the libraries that are being used.

```python
import pandas as pd
import datetime as dt
from neuralprophet import NeuralProphet
```

Load data from AirtrafficA4.csv files and prepare input and output variables

```python
data = pd.read_csv('AirtrafficA4.csv')
year = data['YEAR']
month = data['MONTH']
monthmap = {'JAN':0,'FEB':1,'MAR':2,'APR':3,'MAY':4,'JUNE':5,'JULY':6,'AUG':7,'SEP':8,'OCT':9,'NOV':10,'DEC':11}
passengercounts = data['PASSENGERS CARRIED'].to_numpy()
i=0
time=[]
while i < 128:
    passengercounts[i]=str(str(passengercounts[i]).replace(",",""))
    if passengercounts[i]=='nan':
        passengercounts[i]='0'
    passengercounts[i]=int(passengercounts[i])
    time.append(dt.date(year[i],1+monthmap[month[i].replace(" ","")],1))
    i=i+1

df = pd.DataFrame({
    'ds': time,
    'y': passengercounts
})
```

Implementing Global Model - Neural Prophet

```python
model = NeuralProphet(
    yearly_seasonality=True,
    weekly_seasonality=False,
    daily_seasonality=False,
    n_forecasts=12,
    learning_rate=0.01)
model.fit(df,freq='M')
# Create future DataFrame for predictions
future = model.make_future_dataframe(df,periods=12)  # Predicting for the next 12 months

# Make predictions
forecast = model.predict(future)
```

Saving the predictions in submission1.csv

```python


# Print forecasted values

forecast[['ds', 'yhat1']].to_csv('submission1.csv', index=False)
```
